## Project Objective
Analyze sales data to identify KPIs, trends, regional and category performance, customer segments, risks and business opportunities.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n = 120
dates = pd.date_range('2025-01-01', periods=12, freq='MS')
df = pd.DataFrame({
    'Order_ID': [f'ORD{i:04d}' for i in range(1, n+1)],
    'Date': np.random.choice(dates, n),
    'Region': np.random.choice(['North','South','East','West'], n),
    'Category': np.random.choice(['Electronics','Furniture','Office Supplies'], n),
    'Customer_Segment': np.random.choice(['Consumer','Corporate','Home Office'], n),
    'Sales': np.random.randint(500, 15000, n),
    'Profit': np.random.randint(50, 3500, n)
})
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df.head()
```

## Executive KPIs

```python
revenue = df['Sales'].sum()
orders = df['Order_ID'].nunique()
aov = revenue / orders
profit = df['Profit'].sum()
profit_margin = profit / revenue * 100
kpis = pd.DataFrame({'KPI':['Revenue','Orders','Average Order Value','Profit','Profit Margin'], 'Value':[revenue,orders,round(aov,2),profit,round(profit_margin,2)]})
kpis
```

## Monthly Revenue Trend

```python
monthly = df.groupby('Month', as_index=False)['Sales'].sum()
monthly

plt.figure(figsize=(10,5))
plt.plot(monthly['Month'], monthly['Sales'], marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Regional Performance

```python
regional = df.groupby('Region').agg(Revenue=('Sales','sum'), Profit=('Profit','sum')).sort_values('Revenue', ascending=False)
regional
```

## Product Category Performance

```python
category = df.groupby('Category').agg(Revenue=('Sales','sum'), Profit=('Profit','sum')).sort_values('Revenue', ascending=False)
category
```

## Customer Segment Analysis

```python
segment = df.groupby('Customer_Segment').agg(Revenue=('Sales','sum'), Profit=('Profit','sum'), Orders=('Order_ID','nunique')).sort_values('Revenue', ascending=False)
segment
```

## Risk Identification

- Monitor low-profit regions and categories.
- Investigate months with declining revenue.
- Track customer segments with low contribution.

## Opportunity Identification

- Promote high-margin product categories.
